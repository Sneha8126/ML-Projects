# Spam Mail Prediction

**Import Dependancy**

```python
# Import Dependancy
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.feature_extraction.text import TfidfVectorizer
```

**Import the mail data**

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Toppers Acadmy Data/data/Machine Learning/Chapter 5:- Natural language processing(NLP)/Project/Copy of mail_data.csv')
```

```python
df.head()
```

```python
df['Category'].value_counts()
```

```python
df['Category'] = np.where(df['Category'] == 'ham',1,0)
```

```python
df['Category'].value_counts()
```

### Train Test Split

**Sepret data as x and y**

```python
x = df['Message']
y = df['Category']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=3)
```

### Feature Extraction

**convert text data into numerical data**

```python
x[0]
```

```python
feature_extraction = TfidfVectorizer(min_df=1, stop_words='english', lowercase=True)
```

```python
x_train = feature_extraction.fit_transform(x_train)
x_test = feature_extraction.transform(x_test)
```

```python
print(x_train[0])
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train, y_train)
```

```python
train_pred = model.predict(x_train)
train_acc = accuracy_score(y_train, train_pred)
print(train_acc)
```

```python
test_pred = model.predict(x_test)
test_acc = accuracy_score(y_test, test_pred)
print(test_acc)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
print(confusion_matrix(y_train, train_pred))
```

```python
print(confusion_matrix(y_test, test_pred))
```

### Building a Predictive System

```python
input_data = ["i am going to delhi."]


input_data = feature_extraction.transform(input_data)

prediction = model.predict(input_data)
print(prediction)

if (prediction[0]==1):
  print('Ham mail')

else:
  print('Spam mail')
```
